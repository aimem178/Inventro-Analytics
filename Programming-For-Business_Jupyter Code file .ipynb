```python
def load_data():
    global products_df, sales_df, customers_df   #the three different global dataframes for different files
   
    #LOADING PRODUCTS FROM THE (products.csv)=====================================================
    if os.path.exists(PRODUCTS_FILE):
        products_df = pd.read_csv(PRODUCTS_FILE)
    elif os.path.exists(EXCEL_FILE):
        products_df = pd.read_excel(EXCEL_FILE, sheet_name='Inventro_Products')
        products_df.to_csv(PRODUCTS_FILE, index=False)
        print(Colors.OKGREEN + "Excel converted to products.csv permanently!" + Colors.ENDC)
    else:
        print(Colors.FAIL + "No data file found!" + Colors.ENDC)
        exit()
    #==============================================================================================
   
    #LOADING SALES FROM THE (sales.csv) =====================================================
    if os.path.exists(SALES_FILE):
        sales_df = pd.read_csv(SALES_FILE)
        sales_df['Date'] = pd.to_datetime(sales_df['Date'], errors='coerce')
        print(Colors.OKGREEN + f"Loaded {len(sales_df)} sales records." + Colors.ENDC)
    else:
        sales_df = pd.DataFrame(columns=[
            'SaleID','Date','Time','ProductID','ProductName','Category',
            'QuantitySold','SellingPrice','TotalSaleAmount','ProfitMade','CustomerPhone'
        ])
        sales_df.to_csv(SALES_FILE, index=False)
        print(Colors.WARNING + "New sales.csv created!" + Colors.ENDC)
    #==============================================================================================
       
    #LOEADING CUSTOMERS FROM THE (customers.csv) ==================================================
    if os.path.exists(CUSTOMERS_FILE):
        customers_df = pd.read_csv(CUSTOMERS_FILE)
    else:
        customers_df = pd.DataFrame(columns=['CustomerID','Name','Phone','TotalSpent','Points'])
        customers_df.to_csv(CUSTOMERS_FILE, index=False)
        print(Colors.WARNING + "New customers.csv created!" + Colors.ENDC)
    #==============================================================================================
```

5- SAVE & BACKUP

```python
def backup_and_save(df, filename):
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    backup_path = os.path.join(BACKUP_DIR, f"{filename[:-4]}_backup_{timestamp}.csv")
    shutil.copy(filename, backup_path)
    df.to_csv(filename, index=False)
```

```python
def add_product():
    global products_df                              # ←←← MUST BE THE VERY FIRST LINE
    
    print(Colors.OKCYAN + "\n➕ Add New Product\n" + Colors.ENDC)
    new_id = products_df['ProductID'].max() + 1    # Now this works because global is above
    name = input("Product Name: ")
    category = input("Category: ")
    supplier = input("Supplier: ")
    cost = float(input("Cost Price: "))
    selling = float(input("Selling Price: "))
    qty = int(input("Initial Quantity: "))
    reorder = int(input("Reorder Level: "))
    
    profit_per_unit = round(selling - cost, 2)
    total_value = round(selling * qty, 2)
    potential_profit = round(profit_per_unit * qty, 2)
    
    new_product = pd.DataFrame([{
        'ProductID': new_id, 'ProductName': name, 'Category': category, 'Supplier': supplier,
        'CostPrice': cost, 'SellingPrice': selling, 'QuantityAvailable': qty,
        'ReorderLevel': reorder, 'ProfitPerUnit': profit_per_unit,
        'TotalValueInStock': total_value, 'PotentialProfit': potential_profit
    }])
    
    products_df = pd.concat([products_df, new_product], ignore_index=True)
    backup_and_save(products_df, PRODUCTS_FILE)
    print(Colors.OKGREEN + f"✅ Product Added! ID: {new_id}" + Colors.ENDC)
```

```python
def record_sale():
    global products_df, sales_df, customers_df      # ←←← FIRST LINE, NO CODE ABOVE THIS
    
    cart = []
    total_sale = 0
    total_profit = 0
    
    print(Colors.BOLD + Colors.OKCYAN + "\n🛒 NEW SALE - LET'S MAKE MONEY 💸\n" + Colors.ENDC)
    
    while True:
        search = input(Colors.OKBLUE + "Scan/Search Product (ID or Name, Name, or 'done' to checkout): " + Colors.ENDC).strip()
        if search.lower() == 'done':
            break
        
        result = products_df[products_df['ProductID'].astype(str).str.contains(search, case=False) | 
                            products_df['ProductName'].str.contains(search, case=False, na=False)]
        
        if result.empty:
            print(Colors.FAIL + "❌ Product not found!" + Colors.ENDC)
            continue
        elif len(result) > 1:
            print(result[['ProductID','ProductName','SellingPrice','QuantityAvailable']])
            pid = int(input(Colors.WARNING + "Enter correct ProductID: " + Colors.ENDC))
            product = products_df[products_df['ProductID'] == pid].iloc[0]
        else:
            product = result.iloc[0]
        
        print(f"\n✅ Selected: {product['ProductName']}")
        print(f"   Price: Rs {product['SellingPrice']} | Available: {product['QuantityAvailable']}")
        
        try:
            qty = int(input("   Quantity: "))
        except:
            print(Colors.FAIL + "Invalid quantity!" + Colors.ENDC)
            continue
        
        if qty <= 0:
            print(Colors.FAIL + "Quantity must be positive!" + Colors.ENDC)
            continue
        if qty > product['QuantityAvailable']:
            print(Colors.FAIL + f"❌ Only {product['QuantityAvailable']} in stock!" + Colors.ENDC)
            continue
        
        cart.append({
            'ProductID': product['ProductID'],
            'ProductName': product['ProductName'],
            'Category': product['Category'],
            'QuantitySold': qty,
            'SellingPrice': product['SellingPrice'],
            'ProfitPerUnit': product['ProfitPerUnit']
        })
        
        item_total = qty * product['SellingPrice']
        item_profit = qty * product['ProfitPerUnit']
        total_sale += item_total
        total_profit += item_profit
        
        print(Colors.OKGREEN + f"   ✓ Added {qty} × {product['ProductName']} | +Rs {item_profit:.2f} profit\n" + Colors.ENDC)
    
    if not cart:
        print(Colors.WARNING + "🛒 Cart is empty. Sale cancelled." + Colors.ENDC)
        return
    
    # Customer handling
    phone = input(Colors.OKBLUE + "\nCustomer Phone (or press Enter to skip loyalty): " + Colors.ENDC).strip()
    customer_name = "Walk-in Customer"
    points_earned = 0
    
    if phone:
        if phone in customers_df['Phone'].values:
            customer_row = customers_df[customers_df['Phone'] == phone].iloc[0]
            customer_name = customer_row['Name']
            current_points = customer_row['Points']
            print(f"Welcome back {customer_name}! | Current Points: {current_points}")
        else:
            customer_name = input("New customer name: ").strip()
            if not customer_name:
                customer_name = "Customer"
            new_customer = pd.DataFrame([{
                'CustomerID': len(customers_df) + 1,
                'Name': customer_name,
                'Phone': phone,
                'TotalSpent': 0,
                'Points': 0
            }])
            customers_df = pd.concat([customers_df, new_customer], ignore_index=True)
            print(f"🎉 New customer {customer_name} added!")
    
    # Update stock
    for item in cart:
        products_df.loc[products_df['ProductID'] == item['ProductID'], 'QuantityAvailable'] -= item['QuantitySold']
        products_df.loc[products_df['ProductID'] == item['ProductID'], 'TotalValueInStock'] = \
            products_df['SellingPrice'] * products_df['QuantityAvailable']
        products_df.loc[products_df['ProductID'] == item['ProductID'], 'PotentialProfit'] = \
            products_df['ProfitPerUnit'] * products_df['QuantityAvailable']
    
    # Record sales
    for item in cart:
        sale_row = {
            'SaleID': len(sales_df) + 1,
            'Date': datetime.now().strftime('%Y-%m-%d'),
            'Time': datetime.now().strftime('%H:%M:%S'),
            'ProductID': item['ProductID'],
            'ProductName': item['ProductName'],
            'Category': item['Category'],
            'QuantitySold': item['QuantitySold'],
            'SellingPrice': item['SellingPrice'],
            'TotalSaleAmount': item['QuantitySold'] * item['SellingPrice'],
            'ProfitMade': item['QuantitySold'] * item['ProfitPerUnit'],
            'CustomerPhone': phone if phone else 'N/A'
        }
        sales_df = pd.concat([sales_df, pd.DataFrame([sale_row])], ignore_index=True)
        sales_df['Date'] = pd.to_datetime(sales_df['Date'])  # ←←← ADD THIS TO FORCE ALL DATES TO TIMESTAMP
    
    # Update customer points (10 points per 100 Rs profit)
    if phone:
        points_earned = int(total_profit // 10)
        customers_df.loc[customers_df['Phone'] == phone, 'TotalSpent'] += total_sale
        customers_df.loc[customers_df['Phone'] == phone, 'Points'] += points_earned
    
    # Save everything with backup
    backup_and_save(products_df, PRODUCTS_FILE)
    backup_and_save(sales_df, SALES_FILE)
    backup_and_save(customers_df, CUSTOMERS_FILE)
    
    # Print beautiful invoice
    print_invoice(cart, total_sale, total_profit, phone, points_earned)
    
    print(Colors.BOLD + Colors.PURPLE + f"\n🎯 SALE COMPLETED SUCCESSFULLY!\nTotal: Rs {total_sale:.2f} | Profit: Rs {total_profit:.2f} | Points Earned: {points_earned}\n" + Colors.ENDC)
```

```python
def print_invoice(cart, total_sale, total_profit, phone, points_earned):
    print(Colors.BOLD + Colors.PURPLE + "\n" + "="*60)
    print(" " * 10 + "🧾 INVENTRO ANALYTICS INVOICE 🧾" + " " * 10)
    print("="*60 + Colors.ENDC)
    
    print(f"Date & Time: {datetime.now().strftime('%d-%m-%Y %H:%M:%S')}")
    
    if phone:
        name = customers_df[customers_df['Phone'] == phone]['Name'].values[0]
        total_points = customers_df[customers_df['Phone'] == phone]['Points'].values[0]
        print(f"Customer: {name} | Phone: {phone}")
        print(f"Loyalty: Total Points {total_points} | Earned This Sale: {points_earned} 🎉")
    else:
        print("Customer: Walk-in Customer (No Loyalty)")
    
    print("-"*60)
    print(f"{'Item':<30} {'Qty':<5} {'Price':<10} {'Amount':<10}")
    print("-"*60)
    
    for item in cart:
        amount = item['QuantitySold'] * item['SellingPrice']
        print(f"{item['ProductName'][:29]:<30} {item['QuantitySold']:<5} {item['SellingPrice']:<10.2f} {amount:<10.2f}")
    
    print("-"*60)
    print(f"{'TOTAL SALE':>47} Rs {total_sale:.2f}")
    print(f"{'TOTAL PROFIT (Internal)':>47} Rs {total_profit:.2f}")
    print("="*60)
    print(Colors.OKGREEN + "💸 Thank You! Profits Amplified! 💸" + Colors.ENDC)
    print("="*60 + "\n")
    
    # Bonus ASCII art meter for profit (creative feature!)
    profit_meter = int(min(total_profit / 100, 20))  # Scale to 20 bars max
    print(Colors.OKGREEN + "Profit Meter: " + "█" * profit_meter + " " * (20 - profit_meter) + f" (Rs {total_profit:.2f})" + Colors.ENDC)
    print()
```

<summary style="font-size: 24px; font-weight: bold; color: #fdb813; background: #0f2027; padding: 10px; border-radius: 8px;">
9 - PROFIT ANALYSER (THE BONUS FEATURE OF OUR PRODUCT)
</summary>

```python
def profit_analyser():
    if sales_df.empty:
        print("No sales yet!")
        return
    
    print(Colors.BOLD + Colors.PURPLE + """
========== PROFIT ANALYSER ==========
    """ + Colors.ENDC)
    
    total_profit = sales_df['ProfitMade'].sum()
    print(Colors.BOLD + Colors.WARNING + f"💰 TOTAL PROFIT TILL NOW: Rs {total_profit:.2f}" + Colors.ENDC)
    
    # Top 5 products
    top_products = sales_df.groupby(['ProductID','ProductName'])['ProfitMade'].sum().sort_values(ascending=False).head(5)
    print(Colors.BOLD + Colors.OKCYAN + "\n🏆 TOP 5 PROFITABLE PRODUCTS:\n" + Colors.ENDC)
    for idx, profit in top_products.items():
        pid, name = idx
        profit_ratio = int((profit / total_profit) * 50) if total_profit > 0 else 0
        print(f"{name[:35]:<35} | Rs {profit:.2f} | {'█'*profit_ratio} ({profit_ratio*2}%)")
    
    # Category profit
    cat_profit = sales_df.groupby('Category')['ProfitMade'].sum().sort_values(ascending=False)
    print(Colors.BOLD + Colors.OKCYAN + "\n📊 PROFIT BY CATEGORY:\n" + Colors.ENDC)
    for cat, prof in cat_profit.items():
        profit_ratio = int((prof / total_profit) * 40) if total_profit > 0 else 0
        print(f"{cat[:30]:<30} | Rs {prof:.2f} | {'█'*profit_ratio} ({profit_ratio*2}%)")
    
    # Moving Average Forecast (Smart Version - works even with few days)
    daily_profit = sales_df.groupby('Date')['ProfitMade'].sum()
    
    print(Colors.BOLD + Colors.OKCYAN + "\n📈 FORECAST & TRENDS:\n" + Colors.ENDC)
    
    if len(daily_profit) == 0:
        print("No sales data yet for forecasting.")
    elif len(daily_profit) == 1:
        today_profit = daily_profit.iloc[-1]
        print(f"Only 1 day of data → Today's Profit: Rs {today_profit:.2f}\n")
        print(f"Next Day Forecast (optimistic): Rs {today_profit * 1.1:.2f} (10% growth)")
    else:
        # Use available days (minimum 2, max 7)
        window = min(len(daily_profit), 7)
        ma = daily_profit.rolling(window=window).mean().iloc[-1]
        print(f"{window}-Day Moving Average: Rs {ma:.2f}")
        
        growth_rate = daily_profit.pct_change().mean()
        if growth_rate > 0:
            forecast = ma * (1 + growth_rate)
            print(f"Next Day Forecast: Rs {forecast:.2f} (trending up {growth_rate:.1%})")
        else:
            forecast = ma * 1.03  # conservative 3% growth
            print(f"Next Day Forecast (conservative): Rs {forecast:.2f}")
    
    # Potential profit remaining
    potential = products_df['PotentialProfit'].sum()
    print(Colors.BOLD + Colors.WARNING + f"\n💎 POTENTIAL PROFIT (if everything sold): Rs {potential:.2f}" + Colors.ENDC)
```

```python
def visualizations():
    if sales_df.empty:
        print("No sales data for visualization!")
        return
    
    print(Colors.BOLD + Colors.OKCYAN + "\n📈 VISUALIZATIONS\n" + Colors.ENDC)
    print("1. Sales Over Time (Line)")
    print("2. Profit by Category (Pie)")
    print("3. Top 10 Products by Profit (Bar)")
    print("4. Back")
    
    ch = input("Choose: ")
    
    if ch == '1':
        daily_sales = sales_df.groupby('Date')['TotalSaleAmount'].sum()
        
        if len(daily_sales) == 0:
            print("No sales data yet!")
            return
            
        plt.figure(figsize=(12, 6))
        daily_sales.plot(kind='line', marker='o', linewidth=3, markersize=8, color='#ff6b35')
        
        # THIS IS THE KEY FIX:
        plt.xlim(daily_sales.index.min(), daily_sales.index.max())  # Zoom to actual data range
        
        # Make it pretty
        plt.title('Daily Sales Trend', fontsize=18, fontweight='bold', color='#fdb813')
        plt.xlabel('Date', fontsize=12)
        plt.ylabel('Sales Amount (PKR)', fontsize=12)
        plt.grid(True, alpha=0.3)
        plt.xticks(rotation=45)
        plt.tight_layout()
        
        # Bonus: Add value labels on points
        for date, value in daily_sales.items():
            plt.text(date, value + value*0.02, f'Rs {value:,.0f}', 
                    ha='center', fontsize=10, fontweight='bold')
        
        plt.show()
    
    elif ch == '2':
        cat_sales = sales_df.groupby('Category')['ProfitMade'].sum()
        plt.figure(figsize=(10,8))
        cat_sales.plot(kind='pie', autopct='%1.1f%%', startangle=90)
        plt.title('Profit Contribution by Category')
        plt.ylabel('')
        plt.show()
    
    elif ch == '3':
        top10 = sales_df.groupby(['ProductName'])['ProfitMade'].sum().sort_values(ascending=False).head(10)
        top10.plot(kind='barh', color='green')
        plt.title('Top 10 Profitable Products')
        plt.xlabel('Profit (PKR)')
        plt.show()
```

```python
def generate_daily_report():
    today = datetime.now().date()
    today_sales = sales_df[sales_df['Date'].dt.date == today]  # ←←← USE .dt.date FOR PERFECT MATCH
    if today_sales.empty:
        print("No sales today!")
        return
    
    report = f"""
INVENTRO ANALYTICS - DAILY REPORT ({today.strftime('%d-%m-%Y')})
================================================================
Total Sales Amount: Rs {today_sales['TotalSaleAmount'].sum():.2f}
Total Profit: Rs {today_sales['ProfitMade'].sum():.2f}
Total Items Sold: {today_sales['QuantitySold'].sum()}
Low Stock Items: {len(products_df[products_df['QuantityAvailable'] < 10])}
================================================================
TOP 5 PRODUCTS TODAY:
{today_sales.groupby(['ProductName'])['ProfitMade'].sum().sort_values(ascending=False).head(5)}
================================================================
Thank you for using Inventro Analytics! 🔥
    """
    
    with open(f"Daily_Report_{today.strftime('%Y%m%d')}.txt", "w", encoding='utf-8') as f:
        f.write(report)
    
    print(Colors.OKGREEN + f"Report saved as Daily_Report_{today.strftime('%Y%m%d')}.txt 🔥" + Colors.ENDC)
```
